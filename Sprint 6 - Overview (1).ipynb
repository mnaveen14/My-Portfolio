# Sprint 6 – Feature Engineering & Feature Selection - Overview

## Notebook 1 – Introduction to Feature Engineering

| No. | Topic                        | Simple Definition                                                                             |
| --: | ---------------------------- | --------------------------------------------------------------------------------------------- |
|   1 | What is a Feature?           | A feature is an input column that gives information to the ML model.                          |
|   2 | Feature vs Variable          | A variable is any value or column in data, while a feature is a variable used as model input. |
|   3 | Feature vs Target            | Features are used to make predictions, while the target is what we want to predict.           |
|   4 | Feature Engineering          | Creating or changing features to make them more useful for ML.                                |
|   5 | Why Feature Engineering?     | It helps the model learn useful patterns from the data.                                       |
|   6 | Domain Knowledge             | Understanding the business helps us create meaningful features.                               |
|   7 | Good vs Bad Features         | Good features provide useful information; bad features are irrelevant, noisy, or misleading.  |
|   8 | Feature Engineering Workflow | Understand data → create features → transform → select → check leakage.                       |
|   9 | Feature Creation             | Creating a new feature from existing columns.                                                 |
|  10 | Feature Transformation       | Changing an existing feature into a more useful form or scale.                                |
|  11 | Feature Selection            | Choosing the most useful features and removing unnecessary ones.                              |
|  12 | Feature Extraction           | Taking useful information from existing data and converting it into features.                 |
|  13 | Feature Leakage              | When a feature contains information that would not be available at prediction time.           |
|  14 | Role in ML Pipeline          | It converts clean data into meaningful, ML-ready features.                                    |


## Notebook 2 – Numerical Feature Engineering

| No. | Topic                        | Simple Definition                                                            |
| --: | ---------------------------- | ---------------------------------------------------------------------------- |
|   1 | Mathematical Transformations | Using mathematical operations on existing numbers to create useful features. |
|   2 | Addition                     | Adding values together to create a total.                                    |
|   3 | Subtraction                  | Finding the difference between two values.                                   |
|   4 | Multiplication               | Multiplying values to create a combined measurement.                         |
|   5 | Division                     | Dividing one value by another to find a per-unit or average value.           |
|   6 | Ratios                       | Comparing two values to understand their relationship.                       |
|   7 | Percentages                  | Showing one value as a percentage of another value.                          |
|   8 | Differences                  | Measuring the gap between two numerical values.                              |
|   9 | Aggregations                 | Summarizing multiple values using sum, mean, minimum, maximum, etc.          |
|  10 | Log Transformation           | Reducing the effect of very large or highly skewed values.                   |
|  11 | Square Root Transformation   | Reducing skewness using a less strong transformation than log.               |
|  12 | Power Transformation         | Changing the scale or distribution of a numerical feature using powers.      |
|  13 | Absolute Difference          | Finding the size of a difference without considering its direction.          |
|  14 | Relative Difference          | Measuring a difference compared with a reference value.                      |
|  15 | Rate Features                | Measuring one value relative to another unit, such as sales per day.         |
|  16 | Normalized Features          | Converting values to a common scale, such as 0 to 1.                         |


## Notebook 3 – Categorical Feature Engineering

| No. | Topic                       | Simple Definition                                                            |
| --: | --------------------------- | ---------------------------------------------------------------------------- |
|   1 | Categorical Variables       | Variables that contain groups or categories instead of continuous numbers.   |
|   2 | Nominal Variables           | Categories with no natural order, such as payment method.                    |
|   3 | Ordinal Variables           | Categories that have a meaningful order, such as Basic → Standard → Premium. |
|   4 | Binary Variables            | Variables with only two categories, such as Yes/No.                          |
|   5 | One-Hot Encoding            | Creates separate 0/1 columns for each category.                              |
|   6 | Ordinal Encoding            | Converts ordered categories into numbers while keeping their order.          |
|   7 | Label Encoding              | Assigns numerical labels to categories, commonly used for target variables.  |
|   8 | Frequency Encoding          | Replaces each category with how frequently it appears.                       |
|   9 | Target Encoding             | Replaces a category with a target-based value, such as average churn rate.   |
|  10 | Rare Category Grouping      | Combines categories with very few records into an `Other` group.             |
|  11 | Category Cardinality        | The number of unique categories in a feature.                                |
|  12 | Handling Unknown Categories | Making sure new categories do not cause errors during prediction.            |
|  13 | Combining Categories        | Grouping related categories into a broader category.                         |


## Important Review Questions

| Question                                  | Simple Answer                                                                           |
| ----------------------------------------- | --------------------------------------------------------------------------------------- |
| Why do we need Feature Engineering?       | To create useful information from existing data so the model can learn better patterns. |
| Feature Engineering vs Feature Selection? | Engineering creates or transforms features; selection chooses the useful ones.          |
| Why is domain knowledge important?        | It helps us create features that have real business meaning.                            |
| What is Feature Leakage?                  | Using information that would not be available when making the prediction.               |
| When do we use One-Hot Encoding?          | Mainly for nominal categories with no natural order.                                    |
| When do we use Ordinal Encoding?          | When categories have a meaningful order.                                                |
| What is Cardinality?                      | The number of unique categories in a categorical feature.                               |
| Why use Log Transformation?               | To reduce the effect of very large or skewed values.                                    |
| Why use Normalization?                    | To bring numerical features to a common scale.                                          |
| What is Target Encoding risk?             | It can cause data leakage if target information is not handled correctly.               |
| What is a Good Feature?                   | A feature that provides useful and relevant information for the prediction.             |
| What is a Bad Feature?                    | A feature that is irrelevant, noisy, redundant, or causes leakage.                      |
