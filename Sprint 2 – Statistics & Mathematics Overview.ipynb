## **Notebook 1 – Introduction to Statistics**

| Topic                      | Short & Simple Overview                                                                           |
| -------------------------- | ------------------------------------------------------------------------------------------------- |
| **What is Statistics?**    | The science of collecting, analyzing, interpreting, and presenting data.                          |
| **Importance in AI/ML**    | Statistics helps us understand data and make better predictions and decisions.                    |
| **Statistics vs ML**       | Statistics focuses on understanding data; ML focuses on learning patterns and making predictions. |
| **Descriptive Statistics** | Summarizes existing data using measures like mean, median, and charts.                            |
| **Inferential Statistics** | Uses sample data to make conclusions about a larger population.                                   |
| **Population**             | The complete group being studied. Example: All customers of a company.                            |
| **Sample**                 | A smaller group selected from the population.                                                     |
| **Census**                 | Collecting data from every member of the population.                                              |
| **Sampling**               | Selecting a small group from a larger population.                                                 |
| **Variables**              | Characteristics that can have different values, such as age or salary.                            |
| **Features**               | Input variables used by an ML model.                                                              |
| **Observation**            | One individual record or row in a dataset.                                                        |
| **Dataset**                | A collection of data organized in rows and columns.                                               |
| **Quantitative Data**      | Numerical data, such as age, salary, or marks.                                                    |
| **Qualitative Data**       | Categorical data, such as gender, color, or department.                                           |
| **Continuous Data**        | Can take any value within a range, such as height or weight.                                      |
| **Discrete Data**          | Contains countable values, such as number of students.                                            |
| **Structured Data**        | Organized data in rows and columns, such as Excel or SQL tables.                                  |
| **Unstructured Data**      | Data without a fixed format, such as images, videos, and text.                                    |


## **Notebook 2 – Measures of Central Tendency**

| Topic              | Short & Simple Overview                                        |
| ------------------ | -------------------------------------------------------------- |
| **Mean**           | Average value of all numbers.                                  |
| **Median**         | Middle value after arranging data in order.                    |
| **Mode**           | Most frequently occurring value.                               |
| **Weighted Mean**  | Average where some values have more importance than others.    |
| **Geometric Mean** | Average mainly useful for growth rates and percentage changes. |
| **Harmonic Mean**  | Average useful for rates, such as speed or price ratios.       |


## **Notebook 3 – Measures of Dispersion**

| Topic                        | Short & Simple Overview                                     |
| ---------------------------- | ----------------------------------------------------------- |
| **Range**                    | Difference between the maximum and minimum values.          |
| **Variance**                 | Measures how far data points are spread from the mean.      |
| **Standard Deviation**       | Shows the average spread of values from the mean.           |
| **Quartiles**                | Divide sorted data into four equal parts.                   |
| **Percentiles**              | Divide data into 100 equal parts.                           |
| **IQR**                      | Difference between the third and first quartile: `Q3 - Q1`. |
| **Coefficient of Variation** | Measures relative variability compared with the mean.       |


## **Notebook 4 – Data Distribution**

| Topic                      | Short & Simple Overview                                                          |
| -------------------------- | -------------------------------------------------------------------------------- |
| **Normal Distribution**    | Bell-shaped distribution where most values are around the mean.                  |
| **Uniform Distribution**   | All values have an equal probability of occurring.                               |
| **Gaussian Distribution**  | Another name commonly used for Normal Distribution.                              |
| **Left Skew**              | Long tail on the left side of the distribution.                                  |
| **Right Skew**             | Long tail on the right side of the distribution.                                 |
| **Symmetric Distribution** | Both sides of the distribution are approximately equal.                          |
| **Kurtosis**               | Measures the shape and tail heaviness of a distribution.                         |
| **Empirical Rule**         | In normal data: 68%, 95%, and 99.7% fall within 1, 2, and 3 standard deviations. |


## **Notebook 5 – Probability**

| Topic                           | Short & Simple Overview                                             |
| ------------------------------- | ------------------------------------------------------------------- |
| **Introduction to Probability** | Measures the chance of an event happening.                          |
| **Events**                      | A possible outcome or group of outcomes.                            |
| **Sample Space**                | All possible outcomes of an experiment.                             |
| **Independent Events**          | One event does not affect another event.                            |
| **Dependent Events**            | One event affects the probability of another.                       |
| **Conditional Probability**     | Probability of an event happening given another event has occurred. |
| **Bayes' Theorem**              | Updates probability using new information.                          |
| **Probability Rules**           | Basic rules used to calculate probabilities.                        |
| **Random Variables**            | Variables whose values depend on random outcomes.                   |
| **Expected Value**              | The expected average outcome of a random event.                     |


## **Notebook 6 – Probability Distributions**

| Distribution                 | Short & Simple Overview                                             |
| ---------------------------- | ------------------------------------------------------------------- |
| **Bernoulli Distribution**   | Represents one experiment with two outcomes, such as Yes/No or 0/1. |
| **Binomial Distribution**    | Represents the number of successes in multiple Bernoulli trials.    |
| **Poisson Distribution**     | Models the number of events occurring in a fixed interval.          |
| **Uniform Distribution**     | Every value within a range has an equal probability.                |
| **Normal Distribution**      | Continuous bell-shaped distribution centered around the mean.       |
| **Exponential Distribution** | Models the waiting time between events.                             |


## **Notebook 7 – Sampling**

| Topic                   | Short & Simple Overview                                                      |
| ----------------------- | ---------------------------------------------------------------------------- |
| **Sampling**            | Selecting a smaller group from a population.                                 |
| **Random Sampling**     | Every member has an equal chance of selection.                               |
| **Stratified Sampling** | Population is divided into groups, and samples are selected from each group. |
| **Cluster Sampling**    | Population is divided into clusters, and some clusters are selected.         |
| **Systematic Sampling** | Selecting every `n`th member after a starting point.                         |
| **Sampling Bias**       | When the sample does not properly represent the population.                  |
| **Sampling Error**      | Difference between sample results and actual population results.             |
| **Sample Size**         | The number of observations selected for analysis.                            |


## **Notebook 8 – Correlation & Covariance**

| Topic                    | Short & Simple Overview                                                  |
| ------------------------ | ------------------------------------------------------------------------ |
| **Correlation**          | Measures the strength and direction of a relationship between variables. |
| **Covariance**           | Shows whether two variables change together.                             |
| **Positive Correlation** | Both variables increase or decrease together.                            |
| **Negative Correlation** | One variable increases while the other decreases.                        |
| **No Correlation**       | No meaningful relationship between variables.                            |
| **Pearson Correlation**  | Measures linear relationships between numerical variables.               |
| **Spearman Correlation** | Measures relationships based on ranked values.                           |
| **Correlation Matrix**   | A table showing correlations between multiple variables.                 |
| **Heatmap**              | A color-based visualization of values such as correlations.              |


## **Notebook 9 – Outlier Detection**

| Topic                  | Short & Simple Overview                                                 |
| ---------------------- | ----------------------------------------------------------------------- |
| **What are Outliers?** | Data points that are very different from most other values.             |
| **Types of Outliers**  | Can be global, contextual, or collective outliers.                      |
| **Detecting Outliers** | Identifying unusual values using statistical and visualization methods. |
| **IQR Method**         | Detects values below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR`.        |
| **Z-Score Method**     | Detects values far from the mean using standard deviations.             |
| **Box Plot**           | Visualization useful for identifying outliers and data spread.          |
| **Scatter Plot**       | Helps identify unusual observations and patterns.                       |


## **Notebook 10 – Hypothesis Testing**

| Topic                           | Short & Simple Overview                                                 |
| ------------------------------- | ----------------------------------------------------------------------- |
| **Null Hypothesis (H₀)**        | Assumes there is no significant effect or difference.                   |
| **Alternative Hypothesis (H₁)** | Assumes there is a significant effect or difference.                    |
| **p-value**                     | Measures how compatible the observed data is with the null hypothesis.  |
| **Confidence Interval**         | A range of plausible values for a population parameter.                 |
| **Significance Level (α)**      | Threshold used to decide whether results are statistically significant. |
| **Type I Error**                | Rejecting a true null hypothesis.                                       |
| **Type II Error**               | Failing to reject a false null hypothesis.                              |
| **One-tailed Test**             | Tests for an effect in one specific direction.                          |
| **Two-tailed Test**             | Tests for an effect in either direction.                                |


## **Notebook 11 – Statistical Tests**

| Test                    | Short & Simple Overview                                                                |
| ----------------------- | -------------------------------------------------------------------------------------- |
| **T-Test**              | Compares the means of groups when population variance is unknown.                      |
| **Z-Test**              | Tests a mean or proportion, typically with large samples or known population variance. |
| **Chi-Square Test**     | Tests relationships between categorical variables.                                     |
| **ANOVA**               | Compares the means of three or more groups.                                            |
| **Mann-Whitney U Test** | Non-parametric test for comparing two independent groups.                              |
| **Shapiro-Wilk Test**   | Tests whether data follows a normal distribution.                                      |

## **Notebook 14 – Statistics in Machine Learning**

| ML Area                    | How Statistics is Used                                                    |
| -------------------------- | ------------------------------------------------------------------------- |
| **Data Cleaning**          | Identify incorrect, duplicate, or unusual values.                         |
| **Missing Value Handling** | Use mean, median, mode, or statistical techniques to handle missing data. |
| **Outlier Detection**      | Use IQR, Z-score, and distribution analysis to identify extreme values.   |
| **Feature Engineering**    | Create meaningful features using statistical analysis.                    |
| **Feature Scaling**        | Standardize data using mean and standard deviation.                       |
| **Data Normalization**     | Transform values into a common range or distribution.                     |
| **Model Evaluation**       | Use metrics and statistical analysis to measure model performance.        |
| **Linear Regression**      | Uses statistical relationships to predict continuous values.              |
| **Logistic Regression**    | Uses probability to classify observations into categories.                |
| **Naive Bayes**            | Uses probability and Bayes' theorem for classification.                   |
| **Clustering**             | Uses distance and statistical patterns to group similar observations.     |
| **PCA**                    | Uses variance and covariance to reduce the number of features.            |
| **Recommendation Systems** | Uses statistical patterns and relationships to recommend items.           |
