## **Notebook 1 – NumPy Basics**

| Topic            | Short and Simple Overview                                                                              |
| ---------------- | ------------------------------------------------------------------------------------------------------ |
| NumPy            | A Python library used for working with numerical data and arrays.                                      |
| NumPy Array      | A structure that stores multiple values in a single variable.                                          |
| Creating Arrays  | Arrays can be created using functions like `np.array()`, `np.zeros()`, `np.ones()`, and `np.arange()`. |
| Dimensions       | Arrays can be 1D, 2D, or higher-dimensional.                                                           |
| `ndim`           | Returns the number of dimensions in an array.                                                          |
| `shape`          | Returns the number of rows and columns or the structure of an array.                                   |
| `size`           | Returns the total number of elements in an array.                                                      |
| `dtype`          | Shows the data type of array elements.                                                                 |
| Indexing         | Used to access a specific element using its position.                                                  |
| Slicing          | Used to select a range of elements from an array.                                                      |
| Reshaping        | Changes the structure of an array without changing its data.                                           |
| Flattening       | Converts a multi-dimensional array into a one-dimensional array.                                       |
| Array Operations | Perform mathematical operations like addition, subtraction, multiplication, and division on arrays.    |
| Broadcasting     | Allows operations between arrays with compatible shapes.                                               |


## **Notebook 2 – NumPy Mathematical Operations**

| Topic                 | Short and Simple Overview                                                               |
| --------------------- | --------------------------------------------------------------------------------------- |
| Sum                   | Calculates the total of all values in an array using `np.sum()`.                        |
| Mean                  | Calculates the average value using `np.mean()`.                                         |
| Median                | Finds the middle value after sorting the data using `np.median()`.                      |
| Variance              | Shows how far the values are spread from the mean using `np.var()`.                     |
| Standard Deviation    | Shows the average spread of values from the mean using `np.std()`.                      |
| Minimum               | Finds the smallest value using `np.min()`.                                              |
| Maximum               | Finds the largest value using `np.max()`.                                               |
| Argmin                | Returns the index position of the smallest value using `np.argmin()`.                   |
| Argmax                | Returns the index position of the largest value using `np.argmax()`.                    |
| Percentile            | Finds the value below which a given percentage of data falls using `np.percentile()`.   |
| Dot Product           | Multiplies corresponding values of two arrays and adds the results using `np.dot()`.    |
| Matrix Multiplication | Multiplies two matrices based on rows and columns using `np.matmul()` or `@`.           |
| Transpose             | Changes rows into columns and columns into rows using `.T`.                             |
| Basic Linear Algebra  | Includes operations such as matrix multiplication, transpose, determinant, and inverse. |

# **Notebook 5 – Reading & Writing Data**

| Topic                  | Short and Simple Overview                                                                                                           |
| ---------------------- | ----------------------------------------------------------------------------------------------------------------------------------- |
| CSV Files              | CSV is a simple file format that stores data in rows and columns. Pandas uses `pd.read_csv()` to read it and `to_csv()` to save it. |
| Excel Files            | Excel files can contain one or more sheets. Pandas uses `pd.read_excel()` to read them and `to_excel()` to save data.               |
| JSON Files             | JSON stores data in a structured key-value format. Pandas uses `pd.read_json()` to read JSON data and `to_json()` to save it.       |
| Parquet Files          | Parquet is a file format mainly used for storing large datasets efficiently. Pandas uses `pd.read_parquet()` and `to_parquet()`.    |
| Pickle Files           | Pickle stores Python objects, including Pandas DataFrames. Pandas uses `pd.read_pickle()` and `to_pickle()`.                        |
| SQL Tables             | Data can be read from or written to database tables using Pandas and a database connection.                                         |
| Reading Multiple Files | Multiple files can be read and combined into one DataFrame when the data has the same structure.                                    |
| File Path Handling     | A file path tells Python where a file is stored. It can be a relative path or a full path.                                          |

## **Notebook 14 – Performance Optimization**

| Topic                      | Short and Simple Overview                                                                                                                                                                  |
| -------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------ |
| Vectorization              | Performs an operation on a whole column or array at once instead of using a loop for each value.                                                                                           |
| `apply()` vs Loops         | Both can perform row or value-based operations. `apply()` can make code shorter, while loops process data one by one.                                                                      |
| Memory Optimization        | Reduces the memory used by a DataFrame by selecting suitable data types and avoiding unnecessary data.                                                                                     |
| Category Data Type         | Useful for columns with repeated values, such as department, city, or order status. It can reduce memory usage.                                                                            |
| Efficient Pandas Practices | Includes using vectorization, selecting only required columns, choosing suitable data types, using `query()`, reading large files in chunks, and using efficient file formats when needed. |
