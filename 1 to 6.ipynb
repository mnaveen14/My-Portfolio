| No. | Notebook     | Question                                               | Answer                                                                                                                                                                                          |
| --: | ------------ | ------------------------------------------------------ | ----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
|   1 | Notebook 1   | What is Feature Engineering?                           | Feature Engineering is the process of creating, transforming, or modifying features to give useful information to a Machine Learning model.                                                     |
|   2 | Notebook 1   | Why is Feature Engineering important?                  | It helps the model find useful patterns in the data and can improve model performance.                                                                                                          |
|   3 | Notebook 1   | What is the difference between a Feature and a Target? | A Feature is an input used by the model, while the Target is the output the model tries to predict.                                                                                             |
|   4 | Notebook 1   | What is Feature Creation?                              | Feature Creation means creating a new feature from existing data using a formula or business logic.                                                                                             |
|   5 | Notebook 1   | What is Feature Transformation?                        | Feature Transformation means changing the format or scale of an existing feature to make it more useful for the model.                                                                          |
|   6 | Notebook 1   | What is Feature Leakage?                               | Feature Leakage happens when information that should not be available during prediction gets into the features. It can give unrealistically good model results.                                 |
|   7 | Notebook 1   | Why is Domain Knowledge important?                     | Domain knowledge helps us create features that have real business meaning and are relevant to the problem.                                                                                      |
|   8 | Notebook 2   | Why do we use mathematical transformations?            | They help create new relationships between numerical features and make useful patterns easier for the model to learn.                                                                           |
|   9 | Notebook 2   | When do we use a Ratio feature?                        | We use a ratio when comparing two related values gives more meaningful information, such as spending per order.                                                                                 |
|  10 | Notebook 2   | Why is Log Transformation used?                        | It is mainly used to reduce the effect of highly skewed values and make the distribution more manageable.                                                                                       |
|  11 | Notebook 2   | What is a Normalized Feature?                          | A normalized feature is scaled to a common range so features with different scales can be compared more easily.                                                                                 |
|  12 | Notebook 3   | What is a Categorical Variable?                        | A categorical variable contains groups or labels such as Gender, City, Department, or Payment Method.                                                                                           |
|  13 | Notebook 3   | Nominal vs Ordinal variables?                          | Nominal categories have no natural order, while ordinal categories have a meaningful order.                                                                                                     |
|  14 | Notebook 3   | When do we use One-Hot Encoding?                       | We use it mainly for nominal categories where there is no natural order.                                                                                                                        |
|  15 | Notebook 3   | Ordinal Encoding vs Label Encoding?                    | Ordinal Encoding assigns values based on the actual category order. Label Encoding assigns numbers to categories and does not necessarily represent an order.                                   |
|  16 | Notebook 3   | What is Frequency Encoding?                            | It replaces each category with how frequently that category appears in the dataset.                                                                                                             |
|  17 | Notebook 3   | What is Target Encoding?                               | Target Encoding replaces a category with a statistic based on the target, such as its average target value.                                                                                     |
|  18 | Notebook 3   | What is the risk of Target Encoding?                   | It can cause target leakage if target information from validation or test data is used while creating the encoding.                                                                             |
|  19 | Notebook 4   | Why do we create Date & Time features?                 | Dates contain useful information such as month, weekday, quarter, recency, and seasonality that can help identify patterns.                                                                     |
|  20 | Notebook 4   | What is Recency?                                       | Recency tells us how recently an event happened. For example, how many days since a customer's last purchase.                                                                                   |
|  21 | Notebook 4   | What is Frequency?                                     | Frequency tells us how often an event happens, such as the number of purchases made by a customer.                                                                                              |
|  22 | Notebook 4   | What is Customer Tenure?                               | Customer Tenure is the amount of time a customer has been with the business.                                                                                                                    |
|  23 | Notebook 4   | What is Seasonality?                                   | Seasonality means a pattern that repeats during a specific time period, such as higher sales during holidays.                                                                                   |
|  24 | Notebook 5   | What are Aggregation Features?                         | Aggregation features summarize multiple records using functions such as count, sum, mean, median, minimum, maximum, and standard deviation.                                                     |
|  25 | Notebook 5   | Mean vs Median?                                        | Mean is the average value, while median is the middle value. Median is usually less affected by extreme values.                                                                                 |
|  26 | Notebook 5   | Why use Group-Based Aggregation?                       | It allows us to calculate useful statistics separately for groups such as customers, departments, or products.                                                                                  |
|  27 | Notebook 5   | Give examples of Customer-Level Aggregation.           | Total spending, average spending, number of visits, maximum treatment cost, and standard deviation of spending.                                                                                 |
|  28 | Notebook 5   | Why use Standard Deviation as a feature?               | It shows how much values vary and helps understand the consistency or variation within a group.                                                                                                 |
|  29 | Notebook 6   | What are Interaction Features?                         | Interaction features combine two or more existing features to capture their relationship.                                                                                                       |
|  30 | Notebook 6   | When are Interaction Features useful?                  | They are useful when two features together provide meaningful information that may not be captured separately.                                                                                  |
|  31 | Notebook 6   | Feature × Feature vs Feature / Feature?                | Multiplication captures a combined effect, while division creates a ratio to compare two features.                                                                                              |
|  32 | Notebook 6   | What are Polynomial Features?                          | Polynomial features create powers or combinations of features, such as Age², to capture non-linear relationships.                                                                               |
|  33 | Notebook 6   | Why can too many Interaction Features be a problem?    | Too many interactions can create unnecessary features, increase complexity, cause redundancy, and increase the risk of overfitting.                                                             |
|  34 | Notebook 1–6 | How do you decide whether a feature is useful?         | I check its business meaning, relationship with the target, usefulness for the model, redundancy, and leakage risk.                                                                             |
|  35 | Notebook 1–6 | How do you check for Feature Leakage?                  | I check whether the feature uses target information or information that would only be available after the prediction point.                                                                     |
|  36 | Notebook 1–6 | Explain one feature you created and justify it.        | For example, I created Total_Value by multiplying Quantity and Unit_Price because it represents the total value of the items and provides useful information about the transaction.             |
|  37 | Notebook 1–6 | What is your Feature Engineering workflow?             | I understand the data, identify feature gaps, create useful features, transform features, check their meaning and leakage risk, select useful features, and prepare the final ML-ready dataset. |
