## Notebook 2 - OOPs Concepts

| No. | Topic                    | Easy Understanding                                                                    |
| --- | ------------------------ | ------------------------------------------------------------------------------------- |
| 1   | Class                    | A blueprint for creating objects.                                                     |
| 2   | Object                   | A real instance created from a class.                                                 |
| 3   | Constructor (`__init__`) | Runs automatically when an object is created.                                         |
| 4   | Instance Variables       | Variables unique to each object.                                                      |
| 5   | Class Variables          | Variables shared by all objects.                                                      |
| 6   | Instance Methods         | Methods that work with object data using `self`.                                      |
| 7   | Static Methods           | Methods that don't use object or class data.                                          |
| 8   | Class Methods            | Methods that work with class data using `cls`.                                        |
| 9   | Properties (`@property`) | Access a method like a variable.                                                      |
| 10  | Encapsulation            | Protects data by hiding it inside a class.                                            |
| 11  | Abstraction              | Hides implementation and shows only necessary features.                               |
| 12  | Inheritance              | A child class uses the features of a parent class.                                    |
| 13  | Multiple Inheritance     | One child inherits from multiple parent classes.                                      |
| 14  | Multilevel Inheritance   | Inheritance happens in a chain (Grandparent → Parent → Child).                        |
| 15  | Hierarchical Inheritance | Multiple child classes inherit from one parent class.                                 |
| 16  | Polymorphism             | Same method name gives different behavior for different objects.                      |
| 17  | Method Overriding        | Child class replaces the parent's method.                                             |
| 18  | Method Overloading       | Same method handles different numbers of arguments (using default values or `*args`). |
| 19  | Composition              | One object owns another object.                                                       |
| 20  | Aggregation              | One object uses another object, but both exist independently.                         |
| 21  | Association              | Two objects interact without owning each other.                                       |
| 22  | Magic (Dunder) Methods   | Special methods that define object behavior.                                          |
| 23  | `super()`                | Calls methods or constructors from the parent class.                                  |
| 24  | `__str__()`              | Defines what is shown when an object is printed.                                      |
| 25  | `__repr__()`             | Returns the official representation of an object.                                     |
| 26  | `__len__()`              | Defines what `len()` returns for an object.                                           |
| 27  | Dataclasses              | Automatically creates common methods like `__init__()` and `__repr__()`.              |
| 28  | Singleton Pattern        | Ensures only one object of a class is created.                                        |
